```python
import matplotlib.pyplot as plt
```

```python
# Bridge length
L = float(input("Enter bridge length L (m): "))

# Load
P = float(input("Enter load P (kN): "))

# Position of load from left support
x = float(input("Enter load position x from A (m): "))

print("\nInputs:")
print("Bridge length =", L, "m")
print("Load =", P, "kN")
print("Load position =", x, "m")
```

```python
# Moment equilibrium about A:
# RB * L - P * x = 0

RB = (P * x) / L

# Vertical force equilibrium:
# RA + RB - P = 0

RA = P - RB

print("\nMathematical calculation:")
print("RB = (P × x) / L")
print(f"RB = ({P} × {x}) / {L}")
print(f"RB = {RB:.2f} kN")

print("\nRA = P - RB")
print(f"RA = {P} - {RB:.2f}")
print(f"RA = {RA:.2f} kN")
```

```python
plt.figure(figsize=(10, 4))

# Bridge beam
plt.plot([0, L], [0, 0], linewidth=5)

# Supports
plt.scatter([0, L], [0, 0], s=150)

# Load
plt.arrow(x, 0.8, 0, -0.65,
          head_width=0.25, head_length=0.15,
          length_includes_head=True)

plt.text(x, 0.9, f"Load = {P} kN", ha="center")

# Reaction arrows
plt.arrow(0, -0.8, 0, 0.65,
          head_width=0.25, head_length=0.15,
          length_includes_head=True)

plt.arrow(L, -0.8, 0, 0.65,
          head_width=0.25, head_length=0.15,
          length_includes_head=True)

plt.text(0, -1.0, f"A: RA = {RA:.2f} kN", ha="center")
plt.text(L, -1.0, f"B: RB = {RB:.2f} kN", ha="center")

plt.xlim(-1, L + 1)
plt.ylim(-1.4, 1.3)
plt.xlabel("Position along bridge (m)")
plt.title("Bridge Load and Support Reactions")
plt.grid(True)

plt.show()
```
